# Proyecto FastAPI de Recomendación de Películas

Este proyecto es una API desarrollada con **FastAPI** que provee recomendaciones de películas basadas en un sistema de similitud (usando TF-IDF y similitud de coseno). La API carga y procesa un dataset de películas, realiza transformaciones y expone varios endpoints para consultar información sobre las películas, actores, directores y, en particular, para generar recomendaciones.

---

## Características

- **Carga y ETL:**  
  - Extracción y preprocesamiento del dataset de películas (`movies_final.csv`).
  - Normalización de datos, conversión de fechas y generación de columnas adicionales (por ejemplo, `release_year`, `features`, etc.).

- **Sistema de Recomendación:**  
  - Utiliza **scikit-learn** para vectorizar los datos de características (combinación de géneros, actores y director).
  - Calcula la similitud de coseno para generar recomendaciones basadas en la película consultada.

- **Endpoints de la API:**  
  - Endpoint raíz para verificar que la API esté corriendo.
  - Endpoint `/recomendacion/{titulo}` para obtener una lista de películas recomendadas en función del título ingresado.
  - (Puedes agregar otros endpoints según las necesidades, como consultar detalles de película, votos, etc.)

- **Manejo de Errores:**  
  - Se implementa manejo de errores con `HTTPException` para responder con mensajes claros en caso de título no encontrado o errores internos.
  - Registro de errores en un archivo `error_log.log`.

- **Despliegue en Render:**  
  - La API está preparada para ser desplegada en servicios en la nube como [Render.com](https://render.com/), lo que permite el acceso público y escalabilidad.
